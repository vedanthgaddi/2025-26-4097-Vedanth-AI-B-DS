```python
#12 dataset operations

import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

print("First few records of dataset")
print(df.head())

print("\nTotal bill > 20 and tip > 3")
cond1 = df[(df['total_bill'] > 20) & (df['tip'] > 3)]
print(cond1)

print("\nRecords on Saturday or Sunday")
cond2 = df[(df['day'] == 'Sat') | (df['day'] == 'Sun')]
print(cond2)

print("\nNon smokers")
cond3 = df[df['smoker'] != 'Yes']
print(cond3)

print("\nFemale, non smoker and bill > 15")
cond4 = df[(df['sex'] == 'Female') & (df['smoker'] == 'No') & (df['total_bill'] > 15)]
print(cond4)

print("Count:", len(cond4))

print("\nSorted by total bill")
s1 = df.sort_values('total_bill')
print(s1.head())

print("\nSorted by tip (descending)")
s2 = df.sort_values('tip', ascending=False)
print(s2.head())

print("\nSorted by day and bill")
s3 = df.sort_values(['day','total_bill'])
print(s3.head())

print("\nAverage total bill per day")
g1 = df.groupby('day')['total_bill'].mean()
print(g1)

print("\nAverage tip by sex and smoker")
g2 = df.groupby(['sex','smoker'])['tip'].mean()
print(g2)

print("\nNumber of records by day")
print(df['day'].value_counts())
```

```python
#13 EDA operations

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = [10,12,15,18,20,22,25,30,35,40]

df = pd.DataFrame(data, columns=["values"])

print("Dataset")
print(df)

print("\nBasic statistics")
print(df.describe())

var = np.var(data)
std = np.std(data)

print("\nVariance:", var)
print("Standard Deviation:", std)

plt.hist(data)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram of Data")
plt.show()

mean_val = np.mean(data)
median_val = np.median(data)

print("\nMean:", mean_val)
print("Median:", median_val)

print("\nObservation:")
if mean_val > median_val:
    print("Data is slightly right skewed")
elif mean_val < median_val:
    print("Data is slightly left skewed")
else:
    print("Data looks symmetric")
```

```python
#14 EDA using dataset URL

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/dataprofessor/data/master/BostonHousing.csv"
df = pd.read_csv(url)

print("First 10 rows of dataset")
print(df.head(10))

data = df['rm']

print("\nSummary statistics")
print(data.describe())

print("\nVariance:", np.var(data))
print("Standard deviation:", np.std(data))

plt.hist(data, bins=10)
plt.xlabel("RM (number of rooms)")
plt.ylabel("Frequency")
plt.title("Distribution of RM values")
plt.show()

mean_val = np.mean(data)
median_val = np.median(data)

print("\nMean:", mean_val)
print("Median:", median_val)

if mean_val > median_val:
    print("Data is slightly right skewed")
elif mean_val < median_val:
    print("Data is slightly left skewed")
else:
    print("Data is symmetric")
```

```python
#15 Data Visualization

import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)

print("First 5 rows")
print(df.head())

print("\nDataset info")
df.info()

plt.plot(df['sepal_length'])
plt.title("Sepal Length")
plt.xlabel("Index")
plt.ylabel("Value")
plt.show()

plt.hist(df['petal_length'], bins=20)
plt.title("Petal Length Histogram")
plt.xlabel("Petal Length")
plt.ylabel("Frequency")
plt.show()

plt.scatter(df['sepal_length'], df['petal_length'])
plt.title("Sepal vs Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()

df['species'].value_counts().plot(kind='bar')
plt.title("Species Count")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()

counts = df['species'].value_counts()
plt.pie(counts, labels=counts.index)
plt.title("Species Distribution")
plt.show()
```
